题目：提取含有字符串"Python"的行

```python
import pandas as pd
import numpy as np
data = {"grammer":["Python","C","Java","GO",np.nan,"SQL","PHP","Python"],
       "score":[1,2,np.nan,4,5,6,7,10]}
df = pd.DataFrame(data)
# s1
df[df['grammer']=='Python']

# s2 .str 是 pandas 中的一个访问器，用于字符串类型的列。当你使用 .str
# 访问器时，它允许你执行字符串操作，比如大小写转换、去除空白、分割字符串等
results = df['grammer'].str.contains('Python')
results.fillna(value=False, inplace=True)
df[results]
# results是一个布尔序列，df[results]不能对含有空白行的布尔序列操作，要先用fillna填充
```

题目：将salary列数据转换为最大值与最小值的平均值

```python
import pandas as pd
df = pd.read_excel('pandas120.xlsx')
list = df.salary.str.split('-').to_list()
list_min = [int(x[0].strip('k')) for x in list]
list_max = [int(x[1].strip('k')) for x in list]
list_mean = [int(0.5*(x+y)*1000) for x,y in zip(list_min,list_max)]
df['salary'] = list_mean
df
```

```python
import pandas as pd
df = pd.read_excel('pandas120.xlsx')
def salary_avg(x):
  x_min = int(x.split('-')[0].strip('k'))
  x_max = int(x.split('-')[1].strip('k'))
  return int(0.5*(x_min+x_max)*1000)
df.salary = df.salary.apply(salary_avg)
df.to_excel('pandas120_salary.xlsx', index=False)
df
# 网友解答https://www.heywhale.com/mw/project/604b212d90804e001503e01a
# 该解法的妙处在于df.apply()函数，参数是处理方式
```
