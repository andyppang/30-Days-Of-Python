题目：提取含有字符串"Python"的行

```python
import pandas as pd
import numpy as np
data = {"grammer":["Python","C","Java","GO",np.nan,"SQL","PHP","Python"],
       "score":[1,2,np.nan,4,5,6,7,10]}
df = pd.DataFrame(data)
# s1
df[df['grammer']=='Python']

# s2 .str 是 pandas 中的一个访问器，用于字符串类型的列。当你使用 .str
# 访问器时，它允许你执行字符串操作，比如大小写转换、去除空白、分割字符串等
results = df['grammer'].str.contains('Python')
results.fillna(value=False, inplace=True)
df[results]
# results是一个布尔序列，df[results]不能对含有空白行的布尔序列操作，要先用fillna填充
```

题目：修改第二列列名为'popularity'

```python
import pandas as pd
import numpy as np
data = {"grammer":["Python","C","Java","GO",np.nan,"SQL","PHP","Python"],
       "score":[1,2,np.nan,4,5,6,7,10]}
df = pd.DataFrame(data)
df.columns.values[1] = 'popularity'
df

# df.columns.values: 这是一个NumPy数组（numpy.ndarray），它是df.columns的底层
# 数据结构。当你需要将列名作为数组进行操作时，可以使用.values属性来获取列名的
# 数组表示。这在需要进行数值操作或与其他NumPy数组一起使用时非常有用。
```

```python
import pandas as pd
import numpy as np
data = {"grammer":["Python","C","Java","GO",np.nan,"SQL","PHP","Python"],
       "score":[1,2,np.nan,4,5,6,7,10]}
df = pd.DataFrame(data)
df.rename(columns={'score':"popularity"},inplace=True)
df
```
